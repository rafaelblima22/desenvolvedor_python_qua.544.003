# Análise de dataset

A empresa Umbrella Corporation contratou desenvolvedoras python(🐍) para analisar um dataset para verificar o alto número de cancelamentos de contratos dessa empresa, e para entregar uma solução para este problema.

Para isso, a Umbrella Corp. forneceu um *dataset* com os detalhes dos cancelamentos. Os pythonianos (🐍) devem analisar o arquivo utilizando o **Pandas** (🐼).

Vamos começar trazendo todos os dados do *dataset*:

```python
import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")
display(tabela)
```

## Tratamento de dados

Ao ler o *dataset*, percebemos que existem células com valores nulos, que atrapalham a nossa análise, e portanto devem ser retirados da análise.

Vamos elimina-los e re-exibir os valores, mas primeiro, precisaremos eliminar a coluna **ID** e re-exibir a tabela

```python
tabela = tabela.drop("CustomerID", axis =1)
display(tabela)
```

Vamos pedir para que o python nos retorne o número de linhas preenchidas da tabela:

```python
display(tabela.info())
```

Veja que temos em cada uma das colunas um número diferente de linhas. Isso acontece pq temos dados nulos na tabela, e que devemos eliminar. Portanto,vamos elimina-los atrávez do comando python abaixo e re-exibir a tabela:

```python
tabela = tabela.dropna()
display(tabela.info())
```

Agora sim é que começa a análise de dados propriamente dita. Vamos começar verificando o número de clientes que cancelaram o contrato e o número de clientes que não cancelaram.

```python
display(tabela["cancelou"].value_counts())
```

Podemos perceber que o número de clientes que cancelaram (1.0) realmente é alto, inclusive mais alto que o número de clientes que não cancelaram, mas quanto que isso dá em porcentagem?
Vamos descobrir:

```python
display(tabela["cancelou"].value_counts(normalize=True).map("{:.2%}".format))
```
