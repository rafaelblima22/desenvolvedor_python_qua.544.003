Hoje vamos explorar o uso de Células Python em nossos notebooks

As células de código são blocos interativos onde você pode escrever, testar e executar códigos em Python diretamente no seu ambiente de trabalho, unindo texto explicativo e programação em um só lugar

Observe o exemplo na célula abaixo:

```python
nome = "Rafael Borges"
texto = f"Esta é uma célula Python escrita pelo programador: {nome} 😜"

print(texto)
```

A célula Python tambem permite códigos de entrada de dados. Veja:

```python
nome = input("Informe seu nome: ".strip().title())
print(nome)
```

As células Python são acumulativas, e reconhecem as variáveis das células anteriores.
Observe,por exemplo, a célula abaixo:

```python
n1 = 5
print(f"O valor da variável 'n1' é {n1}.")
```

O valor de n1 é claramente 5. Porém, veja na próxima celula e diga qual sera o valor da variável result:

```python
result = 15 + n1
print(result)
```

O valor de result é 20, pois está somando 15 com n1, que tinha o valor de 5.

Mas agora observe a célula abaixo sem executá-la:

```python
x=10
```

Agora, fça o código na proxima célula, execute e veja qual é o resultado:

```python
result = 25 + x
print(result)
```
